Парсер перебирает страницы выдачи выбранного города и сохраняет каждую карточку в CSV. Для меньшей нагрузки он читает поля из выдачи и не открывает отдельную страницу каждого объявления. Доступные поля: город, марка, цена, год, объём двигателя, состояние, ссылка и краткий фрагмент описания; остальные поля могут быть пустыми.


CSV по умолчанию записывается в папку data с именем kolesa_moto_<city_slug>.csv.
Прогресс сохраняется после каждой страницы, поэтому ячейку можно остановить и запустить снова. При отказе сайта парсер остановится, а не будет пытаться обойти блокировку.
